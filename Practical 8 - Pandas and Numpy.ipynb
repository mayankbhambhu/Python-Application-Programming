```python
import pandas as pd
import numpy as np

# Problem 1: Create a dataframe to display the result from the screenshot
data1 = {
    'Student': ['David', 'Samuel', 'Terry', 'Evan'],
    'Age': [27, 24, 22, 32],
    'Country': ['UK', 'Canada', 'China', 'USA'],
    'Course': ['Python', 'Data Structures', 'Machine Learning', 'Web Development'],
    'Marks': [85, 72, 89, 76]
}
df1 = pd.DataFrame(data1)
print("--- Problem 1: DataFrame ---")
print(df1)

# Problem 2: Retrieve the Marks column and assign it to a variable b
b = df1['Marks']
print("\n--- Problem 2: Variable b (Marks) ---")
print(b)

# Problem 3: Retrieve the Country and Course columns and assign it to a variable c
c = df1[['Country', 'Course']]
print("\n--- Problem 3: Variable c (Country & Course) ---")
print(c)
```

```python
# Problem 4: Data Loading & Exploration

# Creating a dataset with 5 columns and 10 records (including a few NaN values to demonstrate handling)
data2 = {
    'Customer_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Grace', 'Heidi', 'Ivan', 'Judy'],
    'Age': [25, 35, 29, 42, np.nan, 55, 31, 28, 45, 33],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Female', 'Male', 'Female', 'Female', 'Male', 'Female'],
    'Purchase_Amount': [150.50, 200.00, 50.00, 320.75, 120.00, np.nan, 410.20, 90.00, 210.50, 300.00]
}

# Load into Pandas DataFrame
df2 = pd.DataFrame(data2)

# Display the first 5 rows
print("--- First 5 rows using head() ---")
print(df2.head())

# Get summary statistics
print("\n--- Summary statistics using describe() ---")
print(df2.describe())

# Check for missing values
print("\n--- Missing values before handling ---")
print(df2.isnull().sum())

# Handle missing values (Fill Age with median, Purchase_Amount with mean)
df2['Age'] = df2['Age'].fillna(df2['Age'].median())
df2['Purchase_Amount'] = df2['Purchase_Amount'].fillna(df2['Purchase_Amount'].mean())

print("\n--- Missing values after handling ---")
print(df2.isnull().sum())
```

```python
# Problem 5: Data Filtering & Selection

# Retrieve all customers whose Age is greater than 30
age_gt_30 = df2[df2['Age'] > 30]
print("--- Customers older than 30 ---")
print(age_gt_30)

# Find the total Purchase Amount for male and female customers separately
total_purchase_by_gender = df2.groupby('Gender')['Purchase_Amount'].sum()
print("\n--- Total Purchase Amount by Gender ---")
print(total_purchase_by_gender)

# Select only the Customer_ID, Name, and Purchase_Amount columns
selected_columns = df2[['Customer_ID', 'Name', 'Purchase_Amount']]
print("\n--- Selected Columns ---")
print(selected_columns)
```

```python
# Problem 6: Data Transformation & Aggregation

# Add a new column "Discounted_Price" (10% discount on Purchase_Amount)
df2['Discounted_Price'] = df2['Purchase_Amount'] * 0.90
print("--- Dataset with Discounted_Price (First 5 Rows) ---")
print(df2[['Name', 'Purchase_Amount', 'Discounted_Price']].head())

# Group the data by Gender and calculate the average Purchase_Amount
avg_purchase_by_gender = df2.groupby('Gender')['Purchase_Amount'].mean()
print("\n--- Average Purchase Amount by Gender ---")
print(avg_purchase_by_gender)

# Sort the dataset by Purchase_Amount in descending order
sorted_df = df2.sort_values(by='Purchase_Amount', ascending=False)
print("\n--- Dataset sorted by Purchase_Amount (Descending) ---")
print(sorted_df)
```
