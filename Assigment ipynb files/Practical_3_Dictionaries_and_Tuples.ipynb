## Part A - Dictionaries

### 1. Create an empty Dictionary

```python
inventory = {}
```

### 2 & 3. Store the first product details and add to inventory

```python
p1_name = "Mobile Phone"
p1_quantity = 5
p1_price = 20000
p1_release_year = 2020

inventory["Product1"] = {
    "Name": p1_name,
    "Quantity": p1_quantity,
    "Price": p1_price,
    "ReleaseYear": p1_release_year
}
```

### 4 & 5. Store the second product details and add to inventory

```python
p2_name = "Laptop"
p2_quantity = 3
p2_price = 55000
p2_release_year = 2022

inventory["Product2"] = {
    "Name": p2_name,
    "Quantity": p2_quantity,
    "Price": p2_price,
    "ReleaseYear": p2_release_year
}
```

### 6. Display the products present in inventory

```python
print(" Initial Inventory ")
print(inventory)
```

### 7. Check if ReleaseYear is in inventory products

```python
p1_has_year = "ReleaseYear" in inventory["Product1"]
p2_has_year = "ReleaseYear" in inventory["Product2"]

print(f"Product 1 has ReleaseYear: {p1_has_year}")
print(f"Product 2 has ReleaseYear: {p2_has_year}")
```

### 8. Delete release year of both products and print updated inventory

```python
del inventory["Product1"]["ReleaseYear"]
del inventory["Product2"]["ReleaseYear"]

print("--- Inventory After Deleting Release Years ---")
print(inventory)
```

## Part B - Tuples

### 1. Create a tuple called prices

```python
prices = (250, 300, 150, 400, 100, 350, 200)
```

### 2. Find and print the highest and lowest price

```python
highest_price = max(prices)
lowest_price = min(prices)
print(f"Highest Price: {highest_price}")
print(f"Lowest Price: {lowest_price}")
```

### 3. Calculate and print the total sum of all prices

```python
total_sum = sum(prices)
print(f"Total Sum of Prices: {total_sum}")
```

### 4. Convert tuple into a sorted list (ascending order)

```python
sorted_prices = sorted(prices)
print(f"Sorted List of Prices: {sorted_prices}")
```

### 5. Try to modify an element in the tuple (Demonstrate immutability)

```python
try:
    prices[0] = 500  # Attempting to change the first element
except TypeError as e:
    print(f"Modification Error: {e}")
```
