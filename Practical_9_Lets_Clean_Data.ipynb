```python
# Python Assignment: File Operations and String Formatting

##This notebook contains two problems:
##1. `zipper_file()`: Interleaving strings and writing them to a file.
##2. `reformat_date()`: Parsing and reformatting date strings.
```

```python
# Helper function for Problem 1
def zip_strings(s1, s2):
    """
    Zips two strings together character by character.
    If one string is longer, the remaining characters are appended to the end.
    """
    result = []
    min_len = min(len(s1), len(s2))

    # Interleave characters
    for i in range(min_len):
        result.append(s1[i])
        result.append(s2[i])

    # Append any remaining characters from the longer string
    result.append(s1[min_len:])
    result.append(s2[min_len:])

    return "".join(result)

def zipper_file(filename, string_list):
    """
    Takes a filename and a list of strings (even length).
    Writes to the file where each line is a zippered string of sequential pairs.
    """
    with open(filename, 'w') as file:
        for i in range(0, len(string_list), 2):
            str1 = string_list[i]
            str2 = string_list[i+1]
            zippered = zip_strings(str1, str2)
            file.write(zippered + '\n')
```

```python
## Problem 2: Reformat date()

##Brad needs help converting dates from `"MM/DD/YYYY"` format into readable English sentences.
##The output format must be strictly: `"Office Hours on Month DD, YYYY"`.
```

```python
def reformat_date(date_list):
    """
    Takes a list of strings in "MM/DD/YYYY" format and converts them into
    "Office Hours on Month DD, YYYY" format.
    """
    month_mapping = {
        "00": "Unknown", # Handled based on prompt description
        "01": "January",
        "02": "February",
        "03": "March",
        "04": "April",
        "05": "May",
        "06": "June",
        "07": "July",
        "08": "August",
        "09": "September",
        "10": "October",
        "11": "November",
        "12": "December"
    }

    formatted_dates = []

    for date_str in date_list:
        # Split the date string by the '/' delimiter
        parts = date_str.split('/')
        if len(parts) == 3:
            month, day, year = parts
            month_name = month_mapping.get(month, "InvalidMonth")

            # Construct the final sentence
            formatted_dates.append(f"Office Hours on {month_name} {day}, {year}")

    return formatted_dates
```

```python
# --- Test Problem 1 ---
print("Testing zipper_file()...")
test_strings = ["abc", "xyz", "hello", "world", "short", "longer"]
zipper_file("output.txt", test_strings)

# Read the file back to verify
with open("output.txt", "r") as f:
    lines = f.readlines()
    print("Contents of output.txt:")
    for line in lines:
        print(line.strip())
# Expected:
# axbycz
# hweolrllod
# slhoonrgter

print("\n--------------------------\n")

# --- Test Problem 2 ---
print("Testing reformat_date()...")
test_dates = [
    "01/15/2023",
    "12/05/2024",
    "07/31/2025",
    "00/12/2022"
]

results = reformat_date(test_dates)
for res in results:
    print(res)
# Expected:
# Office Hours on January 15, 2023
# Office Hours on December 05, 2024
# Office Hours on July 31, 2025
# Office Hours on Unknown 12, 2022
```
